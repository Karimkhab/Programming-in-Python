```python
import pandas as pd
```

```python
def exchange_seats(seat: pd.DataFrame) -> pd.DataFrame:
    # seat = seat[seat['student']. == (seat['student'].shift(1))]
    new_seat = seat.copy()
    len_seat = len(seat)
    if len_seat%2 != 0:
        len_seat= len_seat-1
    for i in range(0, len_seat, 2):
        new_seat['student'].iloc[i] = seat['student'].iloc[i+1]
        new_seat['student'].iloc[i+1] = seat['student'].iloc[i]
    return new_seat
```

```python
data = [[1, 'Abbot'], [2, 'Doris'], [3, 'Emerson'], [4, 'Green'], [5, 'Jeames']]
seat = pd.DataFrame(data, columns=['id', 'student']).astype({'id':'Int64', 'student':'object'})
```

```python
exchange_seats(seat)
```
