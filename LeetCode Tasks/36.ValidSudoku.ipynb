```python
class Solution(object):
    def isValidSudoku(self, board):
        import numpy as np
        from collections import Counter
        
        board = np.array(board)

        def isvalidarray(arr):
            counts = Counter(arr)
            for i in counts:
                if i == '.':
                    continue
                if counts[i] > 1:
                    return False
            return True

        for i in range(9):
            if not (isvalidarray(board[i, :]) and isvalidarray(board[:, i])):
                return False

        for i in range(0, 9, 3):
            for j in range(0, 9, 3):
                subgrid = board[i:i+3, j:j+3].flatten()
                if not isvalidarray(subgrid):
                    return False

        return True
```
