```python
import pandas as pd

def biggest_single_number(my_numbers: pd.DataFrame) -> pd.DataFrame:
    res = my_numbers.value_counts().reset_index(name='count')
    res = res[res['count']==1].drop(columns=['count'])
    if res.empty: return pd.DataFrame({'num': [None]})
    return res[res['num'] == max(res['num'])]
```

```python
data = [[8], [8], [3], [3]]
my_numbers = pd.DataFrame(data, columns=['num']).astype({'num':'Int64'})
```

```python
biggest_single_number(my_numbers)
```
