```python
import pandas as pd

def tree_node(tree: pd.DataFrame) -> pd.DataFrame:
    tree['Root'], tree['Inner'], tree['type'] = 0, 0, 'Leaf'
    tree.loc[tree['p_id'].isna(), 'Root'] = 1
    tree.loc[tree['id'].isin(tree['p_id'].unique()), 'Inner'] = 1
    
    tree.loc[tree['Inner']==1, 'type'] = 'Inner'
    tree.loc[tree['Root']==1, 'type'] = 'Root'
    tree.drop(columns=['p_id', 'Root', 'Inner'], inplace=True)
    return tree
```

```python
data = [[1, None], [2, 1], [3, 1], [4, 2], [5, 2]]
tree = pd.DataFrame(data, columns=['id', 'p_id']).astype({'id':'Int64', 'p_id':'Int64'})
```

```python
tree_node(tree)
```
