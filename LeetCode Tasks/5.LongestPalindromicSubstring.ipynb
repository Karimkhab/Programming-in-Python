```python
class DataStream(object):

    def __init__(self, value, k):
        """
        :type value: int
        :type k: int
        """
        self.value = value
        self.k = k

        self.counter = 0
        return None

        

    def consec(self, num):
        """
        :type num: int
        :rtype: bool
        """
        if num == self.value: self.counter +=1
        else: self.counter = 0

        if self.counter >= self.k: return True
        return False
```

```python
class Solution(object):
    def maximumWhiteTiles(self, tiles, carpetLen):
        """
        :type tiles: List[List[int]]
        :type carpetLen: int
        :rtype: int
        """
        tiles.sort()
        maximum = 0
        
        for i in range(len(tiles)):
            start, end = tiles[i][0], tiles[i][1]
            carpet_end = start + carpetLen - 1
            
            length = 0
            next_i = i
            
            while next_i < len(tiles) and tiles[next_i][0] <= carpet_end:
                tile_start, tile_end = tiles[next_i][0], tiles[next_i][1]
                
                if tile_end <= carpet_end:
                    length += (tile_end - tile_start + 1)
                else:
                    length += (carpet_end - tile_start + 1)
                    break
                    
                next_i += 1
            
            maximum = max(maximum, length)
        
        return maximum
```

```python
class Solution(object):
    def numGoodSubarrays(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        goods = list()
        for i in range(len(nums)):
            # print('-----------------')
            good = [nums[i]]
            good_sum = nums[i]
            if good_sum%k==0:
                goods.append(good.copy())
                # print('\n========\n',goods, '\n========\n')
            for j in range(i+1, len(nums)):
                # print(goods)
                good.append(nums[j])
                good_sum+=nums[j]
                # print(good, good_sum)
                if good_sum%k==0:
                    goods.append(good.copy())
                    # print('\n========\n',goods, '\n========\n')
        goods_set = set([tuple(item) for item in goods])
        return len(goods_set)
```

```python
class Solution(object):
    def numGoodSubarrays(self, nums, k):
        prefix_mod = 0
        # Словарь: остаток -> set хэшей подмассивов
        mod_hashes = {0: {0}}  # хэш пустого префикса = 0
        result_set = set()
        prefix_hash = 0
        
        for i, num in enumerate(nums):
            prefix_mod = (prefix_mod + num) % k
            prefix_hash = hash((prefix_hash, num))  # хэш текущего префикса
            
            if prefix_mod in mod_hashes:
                # Проверяем все предыдущие префиксы с таким же остатком
                for prev_hash in mod_hashes[prefix_mod]:
                    if prev_hash != prefix_hash:  # если хэши разные, то подмассив разный
                        # Хэш подмассива от (prev_index+1) до i
                        subarray_hash = hash((prev_hash, prefix_hash))
                        result_set.add(subarray_hash)
            
            # Добавляем текущий хэш префикса в множество для этого остатка
            if prefix_mod not in mod_hashes:
                mod_hashes[prefix_mod] = set()
            mod_hashes[prefix_mod].add(prefix_hash)
        
        return len(result_set)
```
